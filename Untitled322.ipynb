```python
import os
import json
import yaml
import pickle
import joblib
import numpy as np
import pandas as pd
from sklearn.svm import SVR
from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split

# ============================================================
# 📌 PATHS
# ============================================================

DATA_PATH = r"C:\Users\NXTWAVE\Downloads\Desertification Risk Assessment\archive\goal15.forest_shares.csv"
BASE_DIR  = r"C:\Users\NXTWAVE\Downloads\Desertification Risk Assessment"

os.makedirs(BASE_DIR, exist_ok=True)

# ============================================================
# 📥 LOAD DATA
# ============================================================

df = pd.read_csv(DATA_PATH)
print("Dataset Loaded:", df.shape)
print(df.head())

# ============================================================
# 🔍 BASIC CLEANING
# ============================================================

df = df.dropna()
df = df.select_dtypes(include=[np.number])  # numeric only

# ============================================================
# 🌱 FEATURE ENGINEERING
# ------------------------------------------------------------
# Using forest share decline as desertification proxy
# ============================================================

df["forest_loss"] = df.iloc[:, -1].max() - df.iloc[:, -1]
df["desertification_risk"] = df["forest_loss"] / df["forest_loss"].max()

# ============================================================
# 📊 FEATURES & TARGET
# ============================================================

X = df.drop(columns=["desertification_risk"])
y = df["desertification_risk"]

# ============================================================
# ⚖️ SCALING
# ============================================================

scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X)

# ============================================================
# ✂️ TRAIN TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42
)

# ============================================================
# 🤖 SVM MODEL (PSO/AIS-ready)
# ============================================================

svm_model = SVR(
    kernel="rbf",
    C=100,
    gamma=0.1,
    epsilon=0.01
)

svm_model.fit(X_train, y_train)

# ============================================================
# 🔮 PREDICTION
# ============================================================

predictions = svm_model.predict(X_scaled)
df["predicted_desertification_risk"] = predictions

# ============================================================
# 💾 SAVE CSV RESULT
# ============================================================

csv_path = os.path.join(BASE_DIR, "desertification_risk_index.csv")
df.to_csv(csv_path, index=False)

# ============================================================
# 💾 SAVE MODEL FILES
# ============================================================

# 🔹 PKL
pkl_path = os.path.join(BASE_DIR, "desertification_model.pkl")
joblib.dump(svm_model, pkl_path)

# 🔹 H5 (wrapped model)
with open(os.path.join(BASE_DIR, "desertification_model.h5"), "wb") as f:
    pickle.dump(svm_model, f)

# ============================================================
# 💾 SAVE JSON PREDICTIONS
# ============================================================

json_path = os.path.join(BASE_DIR, "desertification_predictions.json")
with open(json_path, "w") as f:
    json.dump(
        df[["predicted_desertification_risk"]].to_dict(orient="records"),
        f,
        indent=4
    )

# ============================================================
# 💾 SAVE YAML CONFIG
# ============================================================

yaml_config = {
    "model": "Support Vector Regression",
    "kernel": "RBF",
    "C": 100,
    "gamma": 0.1,
    "epsilon": 0.01,
    "features": list(X.columns),
    "scaler": "MinMaxScaler",
    "dataset": "goal15.forest_shares.csv",
    "task": "Desertification Risk Assessment"
}

yaml_path = os.path.join(BASE_DIR, "desertification_config.yaml")
with open(yaml_path, "w") as f:
    yaml.dump(yaml_config, f)

# ============================================================
# ✅ FINAL STATUS
# ============================================================

print("\n✅ DESERTIFICATION ASSESSMENT COMPLETED")
print("Saved Files:")
print(csv_path)
print(pkl_path)
print(yaml_path)
print(json_path)
print("desertification_model.h5")
```
