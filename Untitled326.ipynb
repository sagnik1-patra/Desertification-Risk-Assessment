```python
import os
import json
import random
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.svm import SVR
from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error

# ============================================================
# 📌 PATHS
# ============================================================

DATA_PATH = r"C:\Users\NXTWAVE\Downloads\Desertification Risk Assessment\archive\goal15.forest_shares.csv"
BASE_DIR  = r"C:\Users\NXTWAVE\Downloads\Desertification Risk Assessment"
os.makedirs(BASE_DIR, exist_ok=True)

# ============================================================
# 🔒 PARAMETER BOUNDS
# ============================================================

C_BOUNDS = (1, 300)
GAMMA_BOUNDS = (0.0001, 1)
EPSILON_BOUNDS = (0.0001, 0.1)

# ============================================================
# 📥 LOAD & PREPROCESS DATA
# ============================================================

df = pd.read_csv(DATA_PATH)
df = df.dropna()
df = df.select_dtypes(include=[np.number])

# ============================================================
# 🌱 FEATURE ENGINEERING
# ============================================================

df["forest_loss"] = df.iloc[:, -1].max() - df.iloc[:, -1]
df["desertification_risk"] = df["forest_loss"] / df["forest_loss"].max()

X = df.drop(columns=["desertification_risk"])
y = df["desertification_risk"]

scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X)

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42
)

# ============================================================
# 🧮 FITNESS FUNCTION
# ============================================================

def fitness(params):
    C, gamma, epsilon = params
    model = SVR(C=C, gamma=gamma, epsilon=epsilon)
    model.fit(X_train, y_train)
    preds = model.predict(X_test)
    return mean_squared_error(y_test, preds)

# ============================================================
# 🧬 AIS (CLONAL SELECTION – GLOBAL SEARCH)
# ============================================================

def AIS(pop_size=12, generations=6):
    population = [
        [
            random.uniform(*C_BOUNDS),
            random.uniform(*GAMMA_BOUNDS),
            random.uniform(*EPSILON_BOUNDS)
        ]
        for _ in range(pop_size)
    ]

    for _ in range(generations):
        population = sorted(population, key=fitness)
        elite = population[:4]

        clones = []
        for p in elite:
            for _ in range(3):
                clone = [
                    np.clip(p[0] * random.uniform(0.9, 1.1), *C_BOUNDS),
                    np.clip(p[1] * random.uniform(0.9, 1.1), *GAMMA_BOUNDS),
                    np.clip(p[2] * random.uniform(0.9, 1.1), *EPSILON_BOUNDS)
                ]
                clones.append(clone)

        population = elite + clones

    return min(population, key=fitness)

# ============================================================
# 🎼 HSA (HARMONY SEARCH – LOCAL REFINEMENT)
# ============================================================

def HSA(init_solution, harmony_size=8, iterations=6, HMCR=0.9, PAR=0.3):
    harmony_memory = [init_solution[:] for _ in range(harmony_size)]

    for _ in range(iterations):
        for i in range(harmony_size):
            new_harmony = []

            for d, bounds in enumerate([C_BOUNDS, GAMMA_BOUNDS, EPSILON_BOUNDS]):
                if random.random() < HMCR:
                    value = harmony_memory[random.randint(0, harmony_size-1)][d]
                    if random.random() < PAR:
                        value += random.uniform(-0.05, 0.05) * value
                else:
                    value = random.uniform(*bounds)

                value = np.clip(value, bounds[0], bounds[1])
                new_harmony.append(value)

            if fitness(new_harmony) < fitness(harmony_memory[i]):
                harmony_memory[i] = new_harmony

    return min(harmony_memory, key=fitness)

# ============================================================
# 🔥 HYBRID AIS → HSA
# ============================================================

ais_best = AIS()
best_params = HSA(ais_best)

C_opt, gamma_opt, epsilon_opt = best_params
print("BEST PARAMETERS (AIS + HSA):", best_params)

# ============================================================
# 🤖 FINAL MODEL
# ============================================================

model = SVR(C=C_opt, gamma=gamma_opt, epsilon=epsilon_opt)
model.fit(X_train, y_train)

y_test_pred = model.predict(X_test)
df["predicted_risk"] = model.predict(X_scaled)

# ============================================================
# 📈 ACCURACY GRAPH
# ============================================================

plt.figure()
plt.plot(y_test.values[:50], label="Actual")
plt.plot(y_test_pred[:50], label="Predicted")
plt.legend()
plt.title("AIS + HSA Accuracy Curve")
plt.show()

plt.savefig(os.path.join(BASE_DIR, "his_accuracy_curve.png"))
plt.close()

# ============================================================
# 📊 COMPARISON GRAPH
# ============================================================

plt.figure()
plt.scatter(y_test, y_test_pred)
plt.xlabel("Actual Risk")
plt.ylabel("Predicted Risk")
plt.title("AIS + HSA Prediction Comparison")
plt.show()

plt.savefig(os.path.join(BASE_DIR, "his_prediction_comparison.png"))
plt.close()

# ============================================================
# 🔥 HEATMAP
# ============================================================

plt.figure(figsize=(10,6))
sns.heatmap(df.corr(), cmap="coolwarm")
plt.title("AIS + HSA Correlation Heatmap")
plt.show()

plt.savefig(os.path.join(BASE_DIR, "his_heatmap.png"))
plt.close()

# ============================================================
# 📉 PREDICTION CURVE
# ============================================================

plt.figure()
plt.plot(df["predicted_risk"])
plt.title("AIS + HSA Desertification Risk Curve")
plt.show()

plt.savefig(os.path.join(BASE_DIR, "his_prediction_curve.png"))
plt.close()

# ============================================================
# 💾 SAVE RESULTS
# ============================================================

csv_path = os.path.join(BASE_DIR, "his_desertification_results.csv")
df.to_csv(csv_path, index=False)

json_path = os.path.join(BASE_DIR, "his_desertification_predictions.json")
with open(json_path, "w") as f:
    json.dump(df[["predicted_risk"]].to_dict(orient="records"), f, indent=4)

# ============================================================
# ✅ COMPLETED
# ============================================================

print("\n✅ AIS + HSA HYBRID DESERTIFICATION MODEL COMPLETED")
print(csv_path)
print(json_path)
```
