```python
# ============================================================
# 🌍 PSO + CSA HYBRID DESERTIFICATION RISK ASSESSMENT (FIXED)
# ============================================================

import os
import json
import random
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.svm import SVR
from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error

# ============================================================
# 📌 PATHS
# ============================================================

DATA_PATH = r"C:\Users\NXTWAVE\Downloads\Desertification Risk Assessment\archive\goal15.forest_shares.csv"
BASE_DIR  = r"C:\Users\NXTWAVE\Downloads\Desertification Risk Assessment"
os.makedirs(BASE_DIR, exist_ok=True)

# ============================================================
# 🔒 PARAMETER BOUNDS (CRITICAL FIX)
# ============================================================

C_BOUNDS = (1, 300)
GAMMA_BOUNDS = (0.0001, 1)
EPSILON_BOUNDS = (0.0001, 0.1)

# ============================================================
# 📥 LOAD & PREPROCESS DATA
# ============================================================

df = pd.read_csv(DATA_PATH)
df = df.dropna()
df = df.select_dtypes(include=[np.number])

# ============================================================
# 🌱 FEATURE ENGINEERING
# ============================================================

df["forest_loss"] = df.iloc[:, -1].max() - df.iloc[:, -1]
df["desertification_risk"] = df["forest_loss"] / df["forest_loss"].max()

X = df.drop(columns=["desertification_risk"])
y = df["desertification_risk"]

scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X)

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42
)

# ============================================================
# 🧮 FITNESS FUNCTION
# ============================================================

def fitness(params):
    C, gamma, epsilon = params
    model = SVR(C=C, gamma=gamma, epsilon=epsilon)
    model.fit(X_train, y_train)
    preds = model.predict(X_test)
    return mean_squared_error(y_test, preds)

# ============================================================
# 🐦 PSO (GLOBAL SEARCH WITH CLIPPING)
# ============================================================

def PSO(particles=10, iterations=6):
    swarm = [
        [
            random.uniform(*C_BOUNDS),
            random.uniform(*GAMMA_BOUNDS),
            random.uniform(*EPSILON_BOUNDS)
        ]
        for _ in range(particles)
    ]

    velocity = [[0, 0, 0] for _ in range(particles)]
    pbest = swarm[:]
    gbest = min(swarm, key=fitness)

    for _ in range(iterations):
        for i in range(particles):
            for d, bounds in enumerate([C_BOUNDS, GAMMA_BOUNDS, EPSILON_BOUNDS]):
                velocity[i][d] = (
                    0.5 * velocity[i][d]
                    + random.random() * (pbest[i][d] - swarm[i][d])
                    + random.random() * (gbest[d] - swarm[i][d])
                )

                swarm[i][d] += velocity[i][d]
                swarm[i][d] = np.clip(swarm[i][d], bounds[0], bounds[1])

            if fitness(swarm[i]) < fitness(pbest[i]):
                pbest[i] = swarm[i]

        gbest = min(pbest, key=fitness)

    return gbest

# ============================================================
# 🐦 CSA (LOCAL REFINEMENT WITH CLIPPING)
# ============================================================

def CSA(init_solution, crows=8, iterations=6, awareness=0.1):
    memory = [init_solution[:] for _ in range(crows)]

    for _ in range(iterations):
        for i in range(crows):
            j = random.randint(0, crows - 1)

            if random.random() > awareness:
                step = random.random()
                new_sol = [
                    memory[i][0] + step * (memory[j][0] - memory[i][0]),
                    memory[i][1] + step * (memory[j][1] - memory[i][1]),
                    memory[i][2] + step * (memory[j][2] - memory[i][2])
                ]
            else:
                new_sol = [
                    random.uniform(*C_BOUNDS),
                    random.uniform(*GAMMA_BOUNDS),
                    random.uniform(*EPSILON_BOUNDS)
                ]

            new_sol[0] = np.clip(new_sol[0], *C_BOUNDS)
            new_sol[1] = np.clip(new_sol[1], *GAMMA_BOUNDS)
            new_sol[2] = np.clip(new_sol[2], *EPSILON_BOUNDS)

            if fitness(new_sol) < fitness(memory[i]):
                memory[i] = new_sol

    return min(memory, key=fitness)

# ============================================================
# 🔥 HYBRID PSO → CSA
# ============================================================

pso_best = PSO()
best_params = CSA(pso_best)

C_opt, gamma_opt, epsilon_opt = best_params
print("BEST PARAMETERS (PSO + CSA):", best_params)

# ============================================================
# 🤖 FINAL MODEL
# ============================================================

model = SVR(C=C_opt, gamma=gamma_opt, epsilon=epsilon_opt)
model.fit(X_train, y_train)

y_test_pred = model.predict(X_test)
df["predicted_risk"] = model.predict(X_scaled)

# ============================================================
# 📈 ACCURACY GRAPH
# ============================================================

plt.figure()
plt.plot(y_test.values[:50], label="Actual")
plt.plot(y_test_pred[:50], label="Predicted")
plt.legend()
plt.title("PSO + CSA Accuracy Curve")
plt.show()

plt.savefig(os.path.join(BASE_DIR, "psa_accuracy_curve.png"))
plt.close()

# ============================================================
# 📊 COMPARISON GRAPH
# ============================================================

plt.figure()
plt.scatter(y_test, y_test_pred)
plt.xlabel("Actual Risk")
plt.ylabel("Predicted Risk")
plt.title("PSO + CSA Prediction Comparison")
plt.show()

plt.savefig(os.path.join(BASE_DIR, "psa_prediction_comparison.png"))
plt.close()

# ============================================================
# 🔥 HEATMAP
# ============================================================

plt.figure(figsize=(10, 6))
sns.heatmap(df.corr(), cmap="coolwarm")
plt.title("PSO + CSA Feature Correlation Heatmap")
plt.show()

plt.savefig(os.path.join(BASE_DIR, "psa_heatmap.png"))
plt.close()

# ============================================================
# 📉 PREDICTION CURVE
# ============================================================

plt.figure()
plt.plot(df["predicted_risk"])
plt.title("PSO + CSA Desertification Risk Curve")
plt.show()

plt.savefig(os.path.join(BASE_DIR, "psa_prediction_curve.png"))
plt.close()

# ============================================================
# 💾 SAVE RESULTS
# ============================================================

csv_path = os.path.join(BASE_DIR, "psa_desertification_results.csv")
df.to_csv(csv_path, index=False)

json_path = os.path.join(BASE_DIR, "psa_desertification_predictions.json")
with open(json_path, "w") as f:
    json.dump(df[["predicted_risk"]].to_dict(orient="records"), f, indent=4)

# ============================================================
# ✅ COMPLETED
# ============================================================

print("\n✅ PSO + CSA HYBRID DESERTIFICATION MODEL COMPLETED")
print(csv_path)
print(json_path)
```
