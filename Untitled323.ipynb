```python
import os
import json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.svm import SVR
from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score

# ============================================================
# 📌 PATHS
# ============================================================

DATA_PATH = r"C:\Users\NXTWAVE\Downloads\Desertification Risk Assessment\archive\goal15.forest_shares.csv"
BASE_DIR  = r"C:\Users\NXTWAVE\Downloads\Desertification Risk Assessment"

os.makedirs(BASE_DIR, exist_ok=True)

# ============================================================
# 📥 LOAD DATA
# ============================================================

df = pd.read_csv(DATA_PATH)
df = df.dropna()
df = df.select_dtypes(include=[np.number])

print("Dataset Shape:", df.shape)

# ============================================================
# 🌱 FEATURE ENGINEERING
# ============================================================

df["forest_loss"] = df.iloc[:, -1].max() - df.iloc[:, -1]
df["desertification_risk"] = df["forest_loss"] / df["forest_loss"].max()

# ============================================================
# 📊 FEATURES & TARGET
# ============================================================

X = df.drop(columns=["desertification_risk"])
y = df["desertification_risk"]

# ============================================================
# ⚖️ SCALING
# ============================================================

scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X)

# ============================================================
# ✂️ SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42
)

# ============================================================
# 🤖 SVM MODEL
# ============================================================

model = SVR(kernel="rbf", C=100, gamma=0.1, epsilon=0.01)
model.fit(X_train, y_train)

# ============================================================
# 🔮 PREDICTIONS
# ============================================================

y_train_pred = model.predict(X_train)
y_test_pred  = model.predict(X_test)

rmse = np.sqrt(mean_squared_error(y_test, y_test_pred))
r2   = r2_score(y_test, y_test_pred)

print(f"RMSE: {rmse:.4f}")
print(f"R2 Score: {r2:.4f}")

# ============================================================
# 📈 1. ACCURACY GRAPH
# ============================================================

plt.figure()
plt.plot(y_test.values[:50], label="Actual")
plt.plot(y_test_pred[:50], label="Predicted")
plt.title("Accuracy Curve (Actual vs Predicted)")
plt.legend()
plt.show()

plt.savefig(os.path.join(BASE_DIR, "accuracy_curve.png"))
plt.close()

# ============================================================
# 📊 2. PREDICTION COMPARISON GRAPH
# ============================================================

plt.figure()
plt.scatter(y_test, y_test_pred)
plt.xlabel("Actual Risk")
plt.ylabel("Predicted Risk")
plt.title("Prediction Comparison")
plt.show()

plt.savefig(os.path.join(BASE_DIR, "prediction_comparison.png"))
plt.close()

# ============================================================
# 🔥 3. HEATMAP
# ============================================================

plt.figure(figsize=(10, 6))
sns.heatmap(df.corr(), cmap="coolwarm")
plt.title("Feature Correlation Heatmap")
plt.show()

plt.savefig(os.path.join(BASE_DIR, "correlation_heatmap.png"))
plt.close()

# ============================================================
# 📉 4. PREDICTION CURVE (FULL DATA)
# ============================================================

df["predicted_risk"] = model.predict(X_scaled)

plt.figure()
plt.plot(df["predicted_risk"])
plt.title("Desertification Risk Prediction Curve")
plt.xlabel("Samples")
plt.ylabel("Risk Index")
plt.show()

plt.savefig(os.path.join(BASE_DIR, "prediction_curve.png"))
plt.close()

# ============================================================
# 💾 SAVE RESULT CSV
# ============================================================

csv_path = os.path.join(BASE_DIR, "desertification_results.csv")
df.to_csv(csv_path, index=False)

# ============================================================
# 💾 SAVE PREDICTION JSON
# ============================================================

json_path = os.path.join(BASE_DIR, "desertification_predictions.json")
with open(json_path, "w") as f:
    json.dump(
        df[["predicted_risk"]].to_dict(orient="records"),
        f,
        indent=4
    )

# ============================================================
# ✅ COMPLETION
# ============================================================

print("\n✅ ALL GRAPHS DISPLAYED AND SAVED")
print("Files Generated:")
print(csv_path)
print(json_path)
print("accuracy_curve.png")
print("prediction_comparison.png")
print("correlation_heatmap.png")
print("prediction_curve.png")
```
