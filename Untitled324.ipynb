```python
import os
import json
import random
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.svm import SVR
from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error

# ============================================================
# 📌 PATHS
# ============================================================

DATA_PATH = r"C:\Users\NXTWAVE\Downloads\Desertification Risk Assessment\archive\goal15.forest_shares.csv"
BASE_DIR  = r"C:\Users\NXTWAVE\Downloads\Desertification Risk Assessment"

os.makedirs(BASE_DIR, exist_ok=True)

# ============================================================
# 📥 LOAD & CLEAN DATA
# ============================================================

df = pd.read_csv(DATA_PATH)
df = df.dropna()
df = df.select_dtypes(include=[np.number])

# ============================================================
# 🌱 FEATURE ENGINEERING
# ============================================================

df["forest_loss"] = df.iloc[:, -1].max() - df.iloc[:, -1]
df["desertification_risk"] = df["forest_loss"] / df["forest_loss"].max()

X = df.drop(columns=["desertification_risk"])
y = df["desertification_risk"]

scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X)

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42
)

# ============================================================
# 🧬 FITNESS FUNCTION
# ============================================================

def fitness(params):
    C, gamma, epsilon = params
    model = SVR(C=C, gamma=gamma, epsilon=epsilon)
    model.fit(X_train, y_train)
    preds = model.predict(X_test)
    return mean_squared_error(y_test, preds)

# ============================================================
# 🧠 AIS (IMMUNE CLONAL SELECTION)
# ============================================================

def AIS(pop_size=10, generations=5):
    population = [
        [random.uniform(10,200), random.uniform(0.001,1), random.uniform(0.001,0.1)]
        for _ in range(pop_size)
    ]

    for _ in range(generations):
        scores = [(fitness(p), p) for p in population]
        scores.sort(key=lambda x: x[0])
        elite = scores[:3]

        clones = []
        for _, p in elite:
            for _ in range(3):
                clone = [
                    p[0] * random.uniform(0.9,1.1),
                    p[1] * random.uniform(0.9,1.1),
                    p[2] * random.uniform(0.9,1.1)
                ]
                clones.append(clone)

        population = [p for _, p in elite] + clones

    best = min(population, key=fitness)
    return best

# ============================================================
# 🐦 PSO (LOCAL REFINEMENT)
# ============================================================

def PSO(init_params, particles=8, iterations=5):
    swarm = [init_params[:] for _ in range(particles)]
    velocities = [[0,0,0] for _ in range(particles)]

    pbest = swarm[:]
    gbest = min(swarm, key=fitness)

    for _ in range(iterations):
        for i in range(particles):
            for d in range(3):
                velocities[i][d] = (
                    0.5 * velocities[i][d]
                    + random.random() * (pbest[i][d] - swarm[i][d])
                    + random.random() * (gbest[d] - swarm[i][d])
                )
                swarm[i][d] += velocities[i][d]

            if fitness(swarm[i]) < fitness(pbest[i]):
                pbest[i] = swarm[i]

        gbest = min(pbest, key=fitness)

    return gbest

# ============================================================
# 🔥 HYBRID AIS → PSO
# ============================================================

ais_best = AIS()
best_params = PSO(ais_best)

C_opt, gamma_opt, epsilon_opt = best_params

print("BEST PARAMETERS:", best_params)

# ============================================================
# 🤖 FINAL MODEL
# ============================================================

model = SVR(C=C_opt, gamma=gamma_opt, epsilon=epsilon_opt)
model.fit(X_train, y_train)

y_test_pred = model.predict(X_test)
df["predicted_risk"] = model.predict(X_scaled)

# ============================================================
# 📈 ACCURACY GRAPH
# ============================================================

plt.figure()
plt.plot(y_test.values[:50], label="Actual")
plt.plot(y_test_pred[:50], label="Predicted")
plt.legend()
plt.title("Hybrid AIS+PSO Accuracy Curve")
plt.show()

plt.savefig(os.path.join(BASE_DIR, "hybrid_accuracy_curve.png"))
plt.close()

# ============================================================
# 📊 COMPARISON GRAPH
# ============================================================

plt.figure()
plt.scatter(y_test, y_test_pred)
plt.xlabel("Actual")
plt.ylabel("Predicted")
plt.title("Hybrid Prediction Comparison")
plt.show()

plt.savefig(os.path.join(BASE_DIR, "hybrid_prediction_comparison.png"))
plt.close()

# ============================================================
# 🔥 HEATMAP
# ============================================================

plt.figure(figsize=(10,6))
sns.heatmap(df.corr(), cmap="coolwarm")
plt.title("Hybrid Feature Correlation Heatmap")
plt.show()

plt.savefig(os.path.join(BASE_DIR, "hybrid_heatmap.png"))
plt.close()

# ============================================================
# 📉 PREDICTION CURVE
# ============================================================

plt.figure()
plt.plot(df["predicted_risk"])
plt.title("Hybrid Desertification Risk Curve")
plt.show()

plt.savefig(os.path.join(BASE_DIR, "hybrid_prediction_curve.png"))
plt.close()

# ============================================================
# 💾 SAVE RESULTS
# ============================================================

csv_path = os.path.join(BASE_DIR, "hybrid_desertification_results.csv")
df.to_csv(csv_path, index=False)

json_path = os.path.join(BASE_DIR, "hybrid_desertification_predictions.json")
with open(json_path, "w") as f:
    json.dump(df[["predicted_risk"]].to_dict(orient="records"), f, indent=4)

# ============================================================
# ✅ DONE
# ============================================================

print("\n✅ HYBRID AIS + PSO DESERTIFICATION MODEL COMPLETED")
print(csv_path)
print(json_path)
```
